# CYB-3 Colab preflight

Select **Runtime → Change runtime type → T4 GPU** if available, then run the check below. It only inspects the runtime; it does not download a model or run a challenge. We need both a GPU and a working Docker daemon before moving the existing experiment to Colab, because the challenge runner depends on Docker isolation.

```python
import shutil
import subprocess

def inspect(command):
    if shutil.which(command[0]) is None:
        print(f'MISSING: {command[0]}')
        return False
    result = subprocess.run(command, text=True, capture_output=True, check=False)
    print(result.stdout or result.stderr)
    print(f'EXIT CODE: {result.returncode}')
    return result.returncode == 0

gpu_ok = inspect(['nvidia-smi'])
docker_ok = inspect(['docker', 'info'])
print('\nPREFLIGHT:', 'GPU_AND_DOCKER_AVAILABLE' if gpu_ok and docker_ok else 'NOT_READY')
```
